```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
print(np.pi)
```

```python
def analytic_pi(max_k):
    k=0
    pi_approx=0
    while k<=max_k:
        pi_approx+=((-1)**k)/(2*k+1)
        k+=1
    return 4*pi_approx
```

```python
print(analytic_pi(1000))
```

```python
def radius_2d(arr):
    r=np.sqrt((arr[0]**2 + arr[1]**2))
    #print(r)
    return r

def MC_estimate_2D(n):
    points=np.random.uniform(0,1,size=(n,2))
    suitable_points=[]
    count=0
    for elt in points:
        #print(elt)
        if radius_2d(elt)<=1:
            suitable_points=np.append(suitable_points,np.array([elt]))
            count+=1
    
        
    # print(suitable_points)
    # print(count)
    return 4*count/n
```

```python
MC_estimate_2D(10)
```

```python
n=1000
start=10
x=np.arange(start,n,4)
y_analytic=np.ones(x.shape)
y_MC=np.ones(x.shape)
y_true=np.pi*np.ones(x.shape)
for i in range (0,len(x)):
    y_analytic[i]=analytic_pi(x[i])
    y_MC[i]=MC_estimate_2D(x[i])
plt.plot(x,y_true, label="Numpy value")
plt.plot(x, y_analytic, label="Analytic approximation")
plt.plot(x, y_MC, label="MC approximation")
plt.xlabel("Number of points chosen/ order k of expansion")
plt.legend()
plt.grid()
```
